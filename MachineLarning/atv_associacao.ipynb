```python
import pandas as pd

from apyori import apriori

base_mercado = pd.read_csv('mercado2.csv', header = None)

transacoes = []

print('tamanho base:', base_mercado[0].count())

for i in range(0, 7501):
    transacoes.append([str(base_mercado.values[i,j]) for j in range(0,20)])

# a) Vamos supor que você quer extrair Regras da Associação para os produtos que são vendidos pelo menos quatro vezes ao dia
# 28 produtos vendidos por semana / 7501 transações
suporte_minimo = 28 / 7501

print('Suporte mínimo:', suporte_minimo)

# b) Adicione o valor de confiança para 0.2 e o lift para 3. Quantos registros foram retornados?
resultados = apriori(
    transacoes,
    min_support=suporte_minimo,
    min_confidence=0.2,
    min_lift=3,
    min_length=2
)


resultados = list(resultados)

# Quantidade de regras encontradas
print('Quantidade de registros retornados:', len(resultados))


# c) Adicione o código abaixo para poder visualizar melhor os dados. Faça comentário do que acontece em cada linha do código.
A = []
B = []
suporte = []
confianca = []
lift = []

for resultado in resultados:

    # suporte da regra
    s = resultado[1]

    # regras encontradas
    result_rules = resultado[2]

    for result_rule in result_rules:

        # antecedente
        a = list(result_rule[0])

        # consequente
        b = list(result_rule[1])

        # confiança
        c = result_rule[2]

        # lift
        l = result_rule[3]

        # adicionando nas listas
        A.append(a)
        B.append(b)
        suporte.append(s)
        confianca.append(c)
        lift.append(l)

# Criando DataFrame com as regras
rules_df = pd.DataFrame({
    'A': A,
    'B': B,
    'suporte': suporte,
    'confianca': confianca,
    'lift': lift
})


# d) Ordene pelo valor de confiança para verificar qual regra tem maior confiabilidade. Qual o valor da regra com maior confiança?
# ORDENAÇÃO POR LIFT
print('\nRegras ordenadas por Lift:\n')

print(
    rules_df.sort_values(
        by='lift',
        ascending=False
    )
)

# ORDENAÇÃO POR CONFIANÇA
print('\nRegra com maior confiança:\n')

print(
    rules_df.sort_values(
        by='confianca',
        ascending=False
    ).head(1)
)
```
